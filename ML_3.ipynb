## Введение в NumPy (Занятие 3)



**NumPy** — библиотека языка Python, позволяющая [удобно] работать с многомерными массивами и матрицами и содержащая математические функции.

## Создание массивов

* Передача итерируемого объекта в качестве параметра функции array (можно также явно указать тип элементов):

```python
q = np.array([0, 10, 1], dtype=float)
q
```

* Создание массивов специального вида при помощи функций:

```python
row = 5
colums = 9
shape = (row, colums)
q = np.ones(shape)
q
```

* Создание последовательностей при помощи функций arange (в качестве парметров принимает левую и правую границы последовательности и **шаг**) и linspace (принимает левую и правую границы и **количество элементов**):

```python
start, stop, step = 0.5, 20.1, 0.4
np.arange(start, stop, step)
```

```python
np.linspace(start, stop)
```

## Базовые операции

* Базовые арифметические операции над массивами выполняются поэлементно.

```python
q = np.array([2, 4, 6])
qq = np.array([5, 8, 10])
q + qq
```

```python
q * qq
```

```python
q / qq
```

```python
q - qq
```

Для выполнения скалярного произведения необходимо использовать функцию dot.

```python
q.dot(qq)
```

## Зачем?

Зачем необходимо использовать NumPy, если существуют стандартные списки/кортежи и циклы?

Причина заключается в скорости работы. Попробуем посчитать сумму поэлементых произведений 2 больших векторов.

```python
import time
```

```python
x = np.random.normal(size = (100_000_000,))
y = np.random.normal(size = (100_000_000,))
x_list = list(x)
y_list = list(y)
```

```python
%%time
sum((x_list[i] * y_list[i] for i in range(100_000_000)))
```

```python
%%time
x @ y
```
