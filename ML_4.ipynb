# Библиотека Matplotlib

## Общее описание
Matplotlib — это мощная библиотека для визуализации данных в Python, предоставляющая широкие возможности для создания статических, анимированных и интерактивных графиков. Она является фундаментальным инструментом для анализа данных и научных исследований.

## Основные возможности

### Типы графиков
Matplotlib поддерживает множество типов визуализаций:
- Линейные графики
- Столбчатые и горизонтальные диаграммы
- Круговые диаграммы
- Гистограммы
- Точечные графики
- Боксплоты
- Графики ошибок
- Контурные графики
- 3D-визуализации

### Ключевые компоненты
1. **Figure (Фигура)** - Контейнер верхнего уровня для всех элементов графика
2. **Axes (Оси)** - Область для построения графиков с системами координат
3. **Axis (Ось)** - Числовая линия с делениями и метками

### Стили и оформление
- Полная настройка цветов, линий и маркеров
- Поддержка LaTeX для математических формул
- Темы и стили оформления
- Кастомизация сетки и легенды

## Преимущества
- Широкая совместимость с другими библиотеками (NumPy, Pandas)
- Гибкость и детальный контроль над каждым аспектом визуализации
- Большое сообщество и обширная документация
- Кроссплатформенная поддержка

## Типичные сценарии использования
- Визуализация результатов научных исследований
- Анализ данных в Data Science
- Создание публикационных графиков
- Разработка интерактивных дашбордов
- Образовательные цели

Matplotlib остается стандартом для визуализации данных в Python, несмотря на появление более современных альтернатив. Его универсальность и мощный функционал делают его стандартным инструментом для работы с данными.

# Импорт библиотек

```python
import csv
import numpy as np
import matplotlib.pyplot as plt
```

# Подготовим данные

Данные получены из Google Trends по запросам 'Python' и 'C++' за период с 2020 по 2025 гг.

```python
py = []
with open('python.csv') as f:
    reader = csv.reader(f)
    for row in reader:
        py.append(int(row[0]))
```

```python
c = []
with open('c++.csv') as f:
    reader = csv.reader(f)
    for row in reader:
        c.append(int(row[0]))
```

# Простейший график

```python
type(py), len(py)
```

```python
type(c), len(c)
```

```python
#Построение графика
y_py = np.array(py)
y_c = np.array(c)
```

```python
x_py = np.arange(0, len(py))
x_c = np.arange(0, len(c))
```

```python
plt.plot(x_py, y_py)
plt.show()
```

```python
plt.plot(x_c, y_c)
plt.plot(x_py, y_py)
plt.show()
```

# Практика 1

Постройте графики функций sin() и cos()
Подсказка: используйте NumPy

```python
start, stop, step = 0, 1080, 1
x_deg = np.arange(start, stop, step)
x = np.radians(x_deg)
y_cos = np.cos(x)
y_sin = np.sin(x)
fig = plt.figure(figsize=(15, 4))
plt.plot(x_deg, y_cos, label='cos')
plt.plot(x_deg, y_sin, label='sin')
plt.xlabel("X")
plt.ylabel("Y")

plt.xticks(np.arange(0, 1081, 90))
plt.legend()
plt.show()
```
