# Регуляризация в линейных моделях

Импортируем библиотеки

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.model_selection import train_test_split

from sklearn.datasets import load_diabetes
```


## Набор данных

**Описание:**
десять исходных переменных, включая возраст, пол, индекс массы тела (BMI), среднее кровяное давление и шесть показателей анализа крови, были собраны для 442 пациентов с диабетом. Также зафиксирована целевая переменная — количественный показатель прогрессирования заболевания через год после начального обследования.

**Характеристики набора данных**
- Количество наблюдений: 442.
- Количество признаков: 10.
- Целевая переменная: количественная мера прогрессирования диабета через год.

**Описание признаков**
1. `age` — возраст в годах.
2. `sex` — пол.
3. `bmi` — индекс массы тела.
4. `bp` — среднее артериальное давление.
5. `s1` — TC, общий холестерин сыворотки.
6. `s2` — LDL, липопротеины низкой плотности.
7. `s3` — HDL, липопротеины высокой плотности.
8. `s4` — TCH, отношение общего холестерина к HDL.
9. `s5` — LTG, уровень триглицеридов.
10. `s6` — GLU, уровень сахара в крови.



Загрузим набор данных

```python
data = load_diabetes()
```

```python
X = pd.DataFrame(data.data, columns = data.feature_names)
y = data.target
X
```

Разделим данные на Train и Test

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

Создадим и обучим модель линейной регрессии

```python
lr = LinearRegression().fit(X_train, y_train)
```

```python
y_pred_train = lr.predict(X_train)
y_pred_test = lr.predict(X_test)
```

```python
mean_squared_error(y_test, y_pred_test) ** 0.5
```

```python
r2_score(y_test, y_pred_test)
```

Коэффициенты LR

```python
lr.coef_[:10]
```

```python
pd.DataFrame(zip(X.columns, lr.coef_.round(5))).sort_values(1)
```
