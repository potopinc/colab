#### Задание 4**.
1. Выбрать все чётные строки матрицы A.
2. Составить одномерный массив из всех не делящихся на 3 элементов нечетных столбцов А.
3. Посчитать сумму диагональных элементов A.

```python
# пример превращения вектора в двумерную матрицу (таблицу)
A = np.arange(120).reshape(10, -1)
A
```

```python
# Выбрать все четные строки матрицы A.
del_ = np.arange(0, A.shape[0], 2)
A = np.delete(A, del_, axis=0)
A
```

```python
# Составить одномерный массив
# из всех не делящихся на 3 элементов нечетных столбцов А.
_del_ = np.arange(0, A.shape[1], 2)
A = np.delete(A, _del_, axis=1)
A.reshape(-1)
A = A[A % 3 != 0]
A
```

```python
A = np.arange(120).reshape(10, -1)
A
```

```python
# Посчитать сумму диагональных элементов A.
np.diagonal(np.fliplr(A)).sum() + np.diagonal(A).sum()
```
