Задание 2.1 - Ответ:

```python
df['month'] = df['production_date'].apply(lambda x: int(x.split('-')[1]))
counts = df['month'].value_counts().sort_index()
counts.plot(kind='bar', color='steelblue', figsize=(10, 5))
plt.title('Количество фильмов по месяцам')
plt.xlabel('Месяц')
plt.ylabel('Количество фильмов')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

Задание 1.1 - Ответ:

```python
df['director_birthYear'] = pd.to_numeric(df['director_birthYear'], errors='coerce')
count = df['director_birthYear'].dropna().astype(int).value_counts().sort_index()
count.plot(kind='bar', color='steelblue', figsize=(12, 5))
plt.title('Количество режиссёров по годам рождения')
plt.xlabel('Год рождения')
plt.ylabel('Количество режиссёров')
plt.grid(axis='y', linestyle='--', alpha=0.2)
plt.tight_layout()
plt.show()
```

Задание 2.2 - Ответ:

```python
df['Year'] = df['production_date'].apply(lambda x: int(x.split('-')[0]))
df['movie_averageRating'].groupby(df['Year']).mean().sort_index().plot(kind='bar', color='steelblue', figsize=(12, 5))
plt.title('Средняя оценка по годам')
plt.xlabel('Год')
plt.ylabel('Оценка')
plt.grid(axis='y', linestyle='--', alpha=0.2)
plt.tight_layout()
plt.show()
```

Задание 2.3 - Ответ:

```python
df['runtime_minutes'] = pd.to_numeric(df['runtime_minutes'], errors='coerce')
df['Year'] = pd.to_numeric(df['production_date'].apply(lambda x: x.split('-')[0] if isinstance(x, str) and '-' in x else None),errors='coerce')
period = df[(df['Year'] >= 1980) & (df['Year'] <= 2020)].copy()
period['year_mean'] = period.groupby('Year')['runtime_minutes'].transform('mean')
period['deviation'] = period['runtime_minutes'] - period['year_mean']
best = period.loc[period['deviation'].idxmax()]

print(f"Год: {int(best['Year'])}")
print(f"Фильм: {best['movie_title']}")
print(f"Длительность: {best['runtime_minutes']:.0f} мин")
print(f"Средняя по году: {best['year_mean']:.1f} мин")
print(f"Отклонение: {best['deviation']:.0f} мин")
```
