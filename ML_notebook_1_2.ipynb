#### **Задание 12**

Напишите функцию для вычисления евклидова расстояния между двумя точками (векторами) одинаковой длины.

Пример: `euclidean_distance([1, 2], [4, 6])` → `5.0`

Расстояние можно посчитать по формуле:

$$
\begin{aligned}
&d(\mathbf{x}, \mathbf{y}) = \sqrt{\sum_{i=1}^{n} (x_i - y_i)^2} \\
&\text{где:} \\
&\quad \mathbf{x} = (x_1, x_2, \dots, x_n) \text{ — координаты первой точки}, \\
&\quad \mathbf{y} = (y_1, y_2, \dots, y_n) \text{ — координаты второй точки},
\end{aligned}
$$

```python
import math
def count(x, y):
  return math.sqrt(sum((xi - yi) ** 2 for xi, yi in zip(x, y)))
print(count([1, 2], [4, 6]))
```

#### **Задание 13**

Реализуйте функцию, которая считает скалярное произведение двух векторов. В качестве векторов функция принимает два списка одинаковой длины.

Скалярное прозведение двух векторов **a** и **b** можно посчитать по формуле:

\begin{aligned}
\mathbf{a} \cdot \mathbf{b} &= \sum_{i=1}^{n} a_i b_i, \\
\text{где:} \\ \mathbf{a} &= (a_1, a_2, \dots, a_n), \\
\mathbf{b} &= (b_1, b_2, \dots, b_n), \\
\end{aligned}


Пример: `dot_product([1, 2, 3], [4, 5, 6])` → `32`

```python
def count(x, y):
  return sum((xi * yi) for xi, yi in zip(x, y))
print(count([1, 2, 3], [4, 5, 6]))
```

#### **Задание 14**

Напишите функцию, которая принимает список чисел и возвращает кортеж из следующих значений:

- максимальное значение,
- минимальное значение,
- среднее значение,
- мода
- медиана

Пример: `stats([1, 2, 3, 4, 4, 5, 5, 5, 6])` → `(6, 1, 3.888888888888889, 5, 4)`

```python
def realize(stats):
  stats = sorted(stats)
  s = list(set(stats))
  c = []
  for num in s:
    c.append(stats.count(num))
  c_of_num = max(c)
  mode = s[c.index(c_of_num)]

  mediana = 0
  if len(stats) % 2 == 0:
    mediana = (stats[len(stats)] + stats[len(stats + 1)]) / 2
  else:
    mediana = stats[len(stats) // 2]

  return (
      max(stats),
      min(stats),
      sum(stats) / len(stats),
      mode,
      mediana
  )
print(realize([1, 2, 3, 4, 4, 5, 5, 5, 6]))
```

#### **Задание 15**

Напишите функцию, которая принимает список чисел и возвращает его масштабированный вариант на интервале `[0, 1]`.

Пример: `normalize([1, 2, 3, 4, 5])` → `[0.0, 0.25, 0.5, 0.75, 1.0]`

```python
def normalize(s):
  dL = round(1 / (max(s) - 1), 2)
  m = []
  for num in s:
    m.append((num - 1) * dL)
  return m
print(normalize([1, 2, 3, 4, 5]))
```
